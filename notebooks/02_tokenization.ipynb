```python
from pathlib import Path

text = Path("../data/processed/lyrics_sample.txt").read_text(
    encoding="utf-8"
)

print(text[:1000])
```

## TODO

Write code that:

1. Calculates average line length
2. Counts unique characters
3. Counts unique words
4. Displays the 20 most common words
5. Reports vocabulary size

```python
lines = text.splitlines()

non_empty_lines = [
    line.strip()
    for line in lines
    if line.strip()
]

average_length = sum(len(line) for line in non_empty_lines) / len(non_empty_lines)

print(f"Average line length: {average_length:.2f}")
```

```python
from collections import Counter

unique_chars = set(text)
print(f"Unique characters: {len(unique_chars):,}")

words = text.lower().split()
unique_words = set(words)
print(f"Unique words: {len(unique_words):,}")

word_counts = Counter(words)
print("\n20 most common words:")
print(word_counts.most_common(20))

print(f"\nVocabulary size: {len(unique_words):,}")
```

```python
from weird_ai.tokenizer import SimpleCharacterTokenizer

tokenizer = SimpleCharacterTokenizer(text)

print(tokenizer.chars)
print(len(tokenizer.chars))
```

```python
encoded = tokenizer.encode("hello class")

print(encoded)
```

```python
decoded = tokenizer.decode(encoded)

print(decoded)
```
